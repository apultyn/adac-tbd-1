### 1.3: Partitioning the dataset

In order to optimize the performance of the queries, we need to partition the dataset. We will use the `ts_date_bucketed` column to partition the dataset.

````python
import shutil
# --- 1.3 Partitioning the dataset -------------------------------------------------
# Path constants used by the benchmark helpers (Part 2) and by Task 2.5.
WORKDIR = Path("data_adac1"); WORKDIR.mkdir(exist_ok=True)
EVENTS_PATH            = str(WORKDIR / "guide_train.parquet")          # default single-file Parquet
OPTIMIZED_EVENTS_PATH  = str(WORKDIR / "guide_train_sorted.parquet")   # sorted on the filter keys + small row groups
PARTITIONED_EVENTS_DIR = str(WORKDIR / "guide_train_by_date")          # Hive-partitioned by ts_date_bucketed
CSV_EVENTS_PATH        = str(WORKDIR / "guide_q1_slice.csv")           # flat CSV negative baseline (only the query's columns)

# (a) one-off CSV -> single Parquet (the dataset ships as CSV; Task 2 / 2.5 read Parquet)
if not os.path.exists(EVENTS_PATH):
    t0 = time.perf_counter()
    pl.scan_csv(train_path, infer_schema_length=20_000).sink_parquet(EVENTS_PATH, compression="zstd")
    print(f"csv -> parquet in {time.perf_counter() - t0:.1f} s")

con = duckdb.connect(); con.execute("SET enable_progress_bar = false")

# (b) derive ts_date_bucketed: the Timestamp dump is heavily front-tailed -- a few thousand rows are
#     scattered across late-2023 / early-2024, then ~99.5% of the rows fall in ~4 weeks of May-Jun 2024.
#     A raw daily Hive partition would create ~150 near-empty directories, so we keep each "dense" day as
#     its own partition and collapse the oldest ~0.5% of rows into one "_other" partition -> ~30 fairly
#     even partitions. (Rationale, metrics and plots: exploratory-analysis/report.html.)
TAIL_FRACTION = 0.005
date_sql   = 'CAST(substr("Timestamp", 1, 10) AS DATE)'
cutoff     = con.execute(f"SELECT quantile_disc({date_sql}, {TAIL_FRACTION}) FROM read_parquet('{EVENTS_PATH}')").fetchone()[0]
bucket_sql = f"CASE WHEN {date_sql} >= DATE '{cutoff}' THEN CAST({date_sql} AS VARCHAR) ELSE '_other' END"
print(f"ts_date_bucketed: dense-day cutoff = {cutoff} (rows older than this date -> '_other')")

# (c) build the three alternative physical layouts with DuckDB COPY
t0 = time.perf_counter()
shutil.rmtree(PARTITIONED_EVENTS_DIR, ignore_errors=True)
con.execute(f"COPY (SELECT *, {bucket_sql} AS ts_date_bucketed FROM read_parquet('{EVENTS_PATH}')) "
            f"TO '{PARTITIONED_EVENTS_DIR}' (FORMAT PARQUET, COMPRESSION ZSTD, PARTITION_BY (ts_date_bucketed), OVERWRITE_OR_IGNORE TRUE)")
con.execute(f"COPY (SELECT * FROM read_parquet('{EVENTS_PATH}') ORDER BY Category, IncidentGrade, CountryCode) "
            f"TO '{OPTIMIZED_EVENTS_PATH}' (FORMAT PARQUET, COMPRESSION ZSTD, ROW_GROUP_SIZE 100000, OVERWRITE_OR_IGNORE TRUE)")
con.execute(f"COPY (SELECT Category, IncidentGrade, CountryCode, IncidentId, MitreTechniques FROM read_parquet('{EVENTS_PATH}')) "
            f"TO '{CSV_EVENTS_PATH}' (FORMAT CSV, HEADER TRUE)")
print(f"built alternative layouts in {time.perf_counter() - t0:.1f} s")

# (d) report sizes + how even the date partitions turned out
pp = con.execute(f"SELECT COUNT(*) n FROM read_parquet('{PARTITIONED_EVENTS_DIR}/**/*.parquet', hive_partitioning=true) GROUP BY ts_date_bucketed").df()["n"]
con.close()
def _dirsize(p): return os.path.getsize(p) if os.path.isfile(p) else sum(os.path.getsize(os.path.join(d, f)) for d, _, fs in os.walk(p) for f in fs)
def _nfiles(p):  return 1 if os.path.isfile(p) else sum(len(fs) for _, _, fs in os.walk(p))
for name, p in [("default parquet (1 file)", EVENTS_PATH), ("sorted parquet (rg=100k)", OPTIMIZED_EVENTS_PATH),
                ("partitioned parquet (by date bucket)", PARTITIONED_EVENTS_DIR), ("csv slice (5 cols, flat)", CSV_EVENTS_PATH)]:
    print(f"  {name:38s} {_dirsize(p)/2**20:9.1f} MB   {_nfiles(p):3d} file(s)")
print(f"  -> {len(pp)} date partitions; rows/partition  min={pp.min():,}  median={int(pp.median()):,}  max={pp.max():,}")
````

````python
import hashlib, threading
import pyarrow.parquet as pq
# --- Part 2: Measuring performance ------------------------------------------------------------------
# Complete benchmark harness -- you do NOT edit this cell. In Tasks 2-5 you call benchmark_case(...) to
# time a query in one library/engine, and show_results() to get the results table. Path constants
# (EVENTS_PATH / PARTITIONED_EVENTS_DIR / OPTIMIZED_EVENTS_PATH / CSV_EVENTS_PATH) come from Part 1.3.
REPEATS = 10  # >= 3 required, 5 recommended (Part 2 protocol); per-call override via benchmark_case(..., repeats=)

BENCHMARK_COLUMNS = [
    "library_engine", "mode", "query_name", "data_format", "layout",
    "rows", "median_time_s", "iqr_time_s", "peak_memory_mb", "input_size_mb",
    "files", "row_groups", "result_check", "notes",
]

def parquet_source(layout):
    if layout == "unpartitioned": return str(EVENTS_PATH)
    if layout == "partitioned":   return str(PARTITIONED_EVENTS_DIR)
    if layout == "optimized":     return str(OPTIMIZED_EVENTS_PATH)
    raise ValueError(layout)

def parquet_glob(layout):
    if layout == "unpartitioned": return str(EVENTS_PATH)
    if layout == "partitioned":   return os.path.join(str(PARTITIONED_EVENTS_DIR), "**", "*.parquet")
    if layout == "optimized":     return str(OPTIMIZED_EVENTS_PATH)
    raise ValueError(layout)

def layout_profile(layout, data_format="parquet"):
    if data_format == "csv":
        return {"files": 1, "row_groups": None, "input_size_mb": round(os.path.getsize(CSV_EVENTS_PATH) / 2**20, 2)}
    path = Path(parquet_source(layout))
    files = sorted(path.rglob("*.parquet")) if path.is_dir() else [path]
    row_groups = sum(pq.ParquetFile(f).num_row_groups for f in files)
    total_bytes = sum(f.stat().st_size for f in files)
    return {"files": len(files), "row_groups": row_groups, "input_size_mb": round(total_bytes / 2**20, 2)}

def to_pandas_result(result):
    if isinstance(result, pd.DataFrame): return result.copy()
    if isinstance(result, pl.DataFrame): return result.to_pandas()
    return pd.DataFrame(result)

def normalize_checksum_value(value):
    if isinstance(value, np.ndarray):        return json.dumps([normalize_checksum_value(v) for v in value.tolist()], sort_keys=True)
    if isinstance(value, (list, tuple)):     return json.dumps([normalize_checksum_value(v) for v in value], sort_keys=True)
    if pd.isna(value):                       return "<NA>"
    if isinstance(value, (np.bool_, bool)):  return "true" if bool(value) else "false"
    if isinstance(value, (np.integer, int)): return str(int(value))
    if isinstance(value, (np.floating, float)):
        value = float(value)
        return str(int(value)) if value.is_integer() else f"{value:.6f}"
    if isinstance(value, (pd.Timestamp, np.datetime64)): return str(pd.Timestamp(value).isoformat())
    return str(value)

def result_checksum(result):
    pdf = to_pandas_result(result).copy()
    pdf.columns = [str(c) for c in pdf.columns]
    for col in pdf.columns: pdf[col] = pdf[col].map(normalize_checksum_value)
    sort_cols = list(pdf.columns)
    if sort_cols: pdf = pdf.sort_values(sort_cols).reset_index(drop=True)
    return hashlib.md5(pdf.to_csv(index=False, lineterminator="\n").encode("utf-8")).hexdigest()[:12]

benchmark_results = []

def run_with_peak_delta(fn, interval=0.01):
    process = psutil.Process(os.getpid())
    def rss_mb(): return process.memory_info().rss / 2**20
    baseline = rss_mb(); peak = [baseline]; running = [True]
    def sample_memory():
        while running[0]:
            peak[0] = max(peak[0], rss_mb()); time.sleep(interval)
    sampler = threading.Thread(target=sample_memory, daemon=True); sampler.start()
    try:
        result = fn(); peak[0] = max(peak[0], rss_mb())
        return max(peak[0] - baseline, 0.0), result
    finally:
        running[0] = False; sampler.join(timeout=1)

def benchmark_case(library_engine, mode, query_name, layout, fn, repeats=REPEATS, notes="", data_format="parquet"):
    times = []; result = None; peak_mem = None
    profile = layout_profile(layout, data_format=data_format)
    for i in range(repeats):
        gc.collect(); start = time.perf_counter()
        if i == 0: peak_mem, result = run_with_peak_delta(fn)
        else:      result = fn()
        times.append(time.perf_counter() - start)
    pdf = to_pandas_result(result)
    row = {
        "library_engine": library_engine, "mode": mode, "query_name": query_name,
        "data_format": data_format, "layout": layout, "rows": len(pdf),
        "median_time_s": round(float(np.median(times)), 4),
        "iqr_time_s": round(float(np.percentile(times, 75) - np.percentile(times, 25)), 4),
        "peak_memory_mb": round(float(peak_mem), 2),
        "input_size_mb": profile["input_size_mb"], "files": profile["files"], "row_groups": profile["row_groups"],
        "result_check": result_checksum(pdf),
        "notes": (notes + "; " if notes else "") + "RSS delta sampled in notebook process",
    }
    benchmark_results.append(row)
    return row

def show_results(filename=None):
    df = pd.DataFrame(benchmark_results).sort_values(["query_name", "layout", "library_engine", "mode"]).reset_index(drop=True)
    if filename:
        os.makedirs("./output", exist_ok=True)
        df.to_csv(f"./output/{filename}", index=False)
    return df

print(f"benchmark harness ready -- {len(BENCHMARK_COLUMNS)} result columns; call benchmark_case(...) / show_results()")
````

### Query 1 - selective filter plus aggregation

```sql
SELECT IncidentGrade, COUNT(*) AS alert_count
FROM events
WHERE Category = 'InitialAccess'
GROUP BY IncidentGrade
ORDER BY alert_count DESC;
```

This query tests how efficiently each engine applies a selective string filter (`Category = 'InitialAccess'`) and then aggregates the results into a small output table. It stresses predicate pushdown and column pruning because only two columns are needed from a 45-column, ~9.5M-row file.

With this query, I expect **DuckDB** to perform best, because it has a cost-based optimizer with vectorized columnar execution and can push the predicate down into the Parquet reader to skip entire row groups whose statistics exclude `'InitialAccess'`. I expect **Pandas (default NumPy backend)** to use the most memory, because it materializes the full dataset before applying any filter, loading all 45 columns into a ~3 GB in-process DataFrame. The physical layout that should help is the **optimized Parquet file sorted by `Category`**, which clusters all `InitialAccess` rows into a contiguous set of row groups and allows the reader to skip the rest entirely.

### Query 2 - query sensitive to partitioned vs. unpartitioned layout

```sql
SELECT ts_date_bucketed, IncidentGrade, COUNT(*) AS alert_count
FROM events
WHERE ts_date_bucketed >= '2024-06-10' AND ts_date_bucketed <= '2024-06-14'
GROUP BY ts_date_bucketed, IncidentGrade
ORDER BY ts_date_bucketed, IncidentGrade;
```

This query tests how well each engine exploits Hive-style partition pruning for a date-range filter. The dataset has 30 date partitions; a 5-day window touches only ~5 of them, so an engine that reads partition metadata before opening files should read roughly 1/6 of the total data compared to a full scan.

With this query, I expect **DuckDB on the partitioned layout** to perform best, because DuckDB resolves the `hive_partitioning` column from the directory names alone and skips the remaining ~25 partition files entirely before reading a single row. I expect **Pandas (default NumPy backend)** to use the most memory, because `read_parquet` on the partitioned directory loads all files into memory and applies the filter afterward. The physical layout that should help is the **Hive-partitioned directory** (`guide_train_by_date/`), which makes the date boundaries visible to the engine as file-system paths and enables file-level pruning without scanning row-group statistics.

Implement your two queries in:

````python
Q2_DATES = ("2024-06-10", "2024-06-14")
Q2_PART_FILTER = [("ts_date_bucketed", ">=", Q2_DATES[0]),
                  ("ts_date_bucketed", "<=", Q2_DATES[1])]

def q1_pd_numpy(path):
    df = pd.read_parquet(path, columns=["Category", "IncidentGrade"])
    return (df[df["Category"] == "InitialAccess"]
            .groupby("IncidentGrade", observed=True).size()
            .reset_index(name="alert_count")
            .sort_values("alert_count", ascending=False).reset_index(drop=True))

def q1_pd_arrow(path):
    df = pd.read_parquet(path, columns=["Category", "IncidentGrade"],
                         engine="pyarrow", dtype_backend="pyarrow")
    return (df[df["Category"] == "InitialAccess"]
            .groupby("IncidentGrade", observed=True).size()
            .reset_index(name="alert_count")
            .sort_values("alert_count", ascending=False).reset_index(drop=True))

def q2_pd_numpy_unpart(path):
    df = pd.read_parquet(path, columns=["Timestamp", "IncidentGrade"])
    df["ts_date"] = df["Timestamp"].str.slice(0, 10)
    mask = (df["ts_date"] >= Q2_DATES[0]) & (df["ts_date"] <= Q2_DATES[1])
    return (df[mask].groupby(["ts_date", "IncidentGrade"], observed=True).size()
            .reset_index(name="alert_count")
            .sort_values(["ts_date", "IncidentGrade"]).reset_index(drop=True))

def q2_pd_arrow_unpart(path):
    df = pd.read_parquet(path, columns=["Timestamp", "IncidentGrade"],
                         engine="pyarrow", dtype_backend="pyarrow")
    df["ts_date"] = df["Timestamp"].str.slice(0, 10)
    mask = (df["ts_date"] >= Q2_DATES[0]) & (df["ts_date"] <= Q2_DATES[1])
    return (df[mask].groupby(["ts_date", "IncidentGrade"], observed=True).size()
            .reset_index(name="alert_count")
            .sort_values(["ts_date", "IncidentGrade"]).reset_index(drop=True))

def q2_pd_numpy_part():
    df = pd.read_parquet(PARTITIONED_EVENTS_DIR,
                         columns=["ts_date_bucketed", "IncidentGrade"],
                         filters=Q2_PART_FILTER)
    return (df.groupby(["ts_date_bucketed", "IncidentGrade"], observed=True).size()
            .reset_index(name="alert_count")
            .sort_values(["ts_date_bucketed", "IncidentGrade"]).reset_index(drop=True))

def q2_pd_arrow_part():
    df = pd.read_parquet(PARTITIONED_EVENTS_DIR,
                         columns=["ts_date_bucketed", "IncidentGrade"],
                         filters=Q2_PART_FILTER,
                         engine="pyarrow", dtype_backend="pyarrow")
    return (df.groupby(["ts_date_bucketed", "IncidentGrade"], observed=True).size()
            .reset_index(name="alert_count")
            .sort_values(["ts_date_bucketed", "IncidentGrade"]).reset_index(drop=True))
````

````python

# ── Task 2: Polars implementations ────────────────────────────────────────────
# Q1: Category filter + IncidentGrade aggregation
# Q2: date-range filter (from Timestamp or ts_date_bucketed) + IncidentGrade aggregation
# Modes: eager | lazy (collect) | lazy streaming (collect(engine="streaming"))

Q2_DATES = ("2024-06-10", "2024-06-14")

# ── Q1: Polars ─────────────────────────────────────────────────────────────────
def q1_pl_eager(path):
    return (pl.read_parquet(path, columns=["Category", "IncidentGrade"])
              .filter(pl.col("Category") == "InitialAccess")
              .group_by("IncidentGrade").agg(pl.len().alias("alert_count"))
              .sort("alert_count", descending=True))

def q1_pl_lazy(path):
    return (pl.scan_parquet(path)
              .select(["Category", "IncidentGrade"])
              .filter(pl.col("Category") == "InitialAccess")
              .group_by("IncidentGrade").agg(pl.len().alias("alert_count"))
              .sort("alert_count", descending=True)
              .collect())

def q1_pl_streaming(path):
    return (pl.scan_parquet(path)
              .select(["Category", "IncidentGrade"])
              .filter(pl.col("Category") == "InitialAccess")
              .group_by("IncidentGrade").agg(pl.len().alias("alert_count"))
              .sort("alert_count", descending=True)
              .collect(engine="streaming"))

for layout in ["unpartitioned", "optimized"]:
    path = parquet_source(layout)
    benchmark_case("polars", "eager",     "Q1_filter_agg", layout, lambda p=path: q1_pl_eager(p))
    benchmark_case("polars", "lazy",      "Q1_filter_agg", layout, lambda p=path: q1_pl_lazy(p))
    benchmark_case("polars", "streaming", "Q1_filter_agg", layout, lambda p=path: q1_pl_streaming(p))

# ── Q2: Polars unpartitioned / optimized (derive date from Timestamp) ──────────
def q2_pl_eager_unpart(path):
    return (pl.read_parquet(path, columns=["Timestamp", "IncidentGrade"])
              .with_columns(pl.col("Timestamp").str.slice(0, 10).alias("ts_date"))
              .filter((pl.col("ts_date") >= Q2_DATES[0]) & (pl.col("ts_date") <= Q2_DATES[1]))
              .group_by(["ts_date", "IncidentGrade"]).agg(pl.len().alias("alert_count"))
              .sort(["ts_date", "IncidentGrade"]))

def q2_pl_lazy_unpart(path):
    return (pl.scan_parquet(path)
              .select(["Timestamp", "IncidentGrade"])
              .with_columns(pl.col("Timestamp").str.slice(0, 10).alias("ts_date"))
              .filter((pl.col("ts_date") >= Q2_DATES[0]) & (pl.col("ts_date") <= Q2_DATES[1]))
              .group_by(["ts_date", "IncidentGrade"]).agg(pl.len().alias("alert_count"))
              .sort(["ts_date", "IncidentGrade"])
              .collect())

def q2_pl_streaming_unpart(path):
    return (pl.scan_parquet(path)
              .select(["Timestamp", "IncidentGrade"])
              .with_columns(pl.col("Timestamp").str.slice(0, 10).alias("ts_date"))
              .filter((pl.col("ts_date") >= Q2_DATES[0]) & (pl.col("ts_date") <= Q2_DATES[1]))
              .group_by(["ts_date", "IncidentGrade"]).agg(pl.len().alias("alert_count"))
              .sort(["ts_date", "IncidentGrade"])
              .collect(engine="streaming"))

for layout in ["unpartitioned", "optimized"]:
    path = parquet_source(layout)
    benchmark_case("polars", "eager",     "Q2_date_range", layout, lambda p=path: q2_pl_eager_unpart(p))
    benchmark_case("polars", "lazy",      "Q2_date_range", layout, lambda p=path: q2_pl_lazy_unpart(p))
    benchmark_case("polars", "streaming", "Q2_date_range", layout, lambda p=path: q2_pl_streaming_unpart(p))

# ── Q2: Polars partitioned (hive_partitioning enables file-level pruning) ───────
PART_GLOB = parquet_glob("partitioned")

def q2_pl_eager_part():
    return (pl.read_parquet(PART_GLOB, hive_partitioning=True)
              .filter((pl.col("ts_date_bucketed") >= Q2_DATES[0]) &
                      (pl.col("ts_date_bucketed") <= Q2_DATES[1]))
              .group_by(["ts_date_bucketed", "IncidentGrade"]).agg(pl.len().alias("alert_count"))
              .sort(["ts_date_bucketed", "IncidentGrade"]))

def q2_pl_lazy_part():
    return (pl.scan_parquet(PART_GLOB, hive_partitioning=True)
              .filter((pl.col("ts_date_bucketed") >= Q2_DATES[0]) &
                      (pl.col("ts_date_bucketed") <= Q2_DATES[1]))
              .group_by(["ts_date_bucketed", "IncidentGrade"]).agg(pl.len().alias("alert_count"))
              .sort(["ts_date_bucketed", "IncidentGrade"])
              .collect())

def q2_pl_streaming_part():
    return (pl.scan_parquet(PART_GLOB, hive_partitioning=True)
              .filter((pl.col("ts_date_bucketed") >= Q2_DATES[0]) &
                      (pl.col("ts_date_bucketed") <= Q2_DATES[1]))
              .group_by(["ts_date_bucketed", "IncidentGrade"]).agg(pl.len().alias("alert_count"))
              .sort(["ts_date_bucketed", "IncidentGrade"])
              .collect(engine="streaming"))

benchmark_case("polars", "eager",     "Q2_date_range", "partitioned", q2_pl_eager_part)
benchmark_case("polars", "lazy",      "Q2_date_range", "partitioned", q2_pl_lazy_part)
benchmark_case("polars", "streaming", "Q2_date_range", "partitioned", q2_pl_streaming_part)
````

````python

# ── Task 2: DuckDB implementations ────────────────────────────────────────────
# DuckDB reads Parquet directly; no need to load into Python first.
# Q2 on unpartitioned/optimized derives the date from Timestamp via substr().
# Q2 on partitioned uses hive_partitioning=true so DuckDB resolves ts_date_bucketed
# from directory names and skips non-matching partition files entirely.

con_q = duckdb.connect(); con_q.execute("SET enable_progress_bar = false")

Q2_DATES = ("2024-06-10", "2024-06-14")

# ── Q1: DuckDB ─────────────────────────────────────────────────────────────────
def q1_duckdb(path):
    return con_q.execute(f"""
        SELECT IncidentGrade, COUNT(*) AS alert_count
        FROM read_parquet('{path}')
        WHERE Category = 'InitialAccess'
        GROUP BY IncidentGrade
        ORDER BY alert_count DESC
    """).df()

for layout in ["unpartitioned", "optimized"]:
    path = parquet_source(layout).replace("\\", "/")
    benchmark_case("duckdb", "sql", "Q1_filter_agg", layout, lambda p=path: q1_duckdb(p))

# ── Q2: DuckDB unpartitioned / optimized (derive date from Timestamp) ──────────
def q2_duckdb_unpart(path):
    return con_q.execute(f"""
        SELECT substr("Timestamp", 1, 10) AS ts_date,
               IncidentGrade,
               COUNT(*) AS alert_count
        FROM read_parquet('{path}')
        WHERE substr("Timestamp", 1, 10) >= '{Q2_DATES[0]}'
          AND substr("Timestamp", 1, 10) <= '{Q2_DATES[1]}'
        GROUP BY ts_date, IncidentGrade
        ORDER BY ts_date, IncidentGrade
    """).df()

for layout in ["unpartitioned", "optimized"]:
    path = parquet_source(layout).replace("\\", "/")
    benchmark_case("duckdb", "sql", "Q2_date_range", layout, lambda p=path: q2_duckdb_unpart(p))

# ── Q2: DuckDB partitioned (hive_partitioning enables file-level pruning) ───────
# DuckDB reads partition column from directory names, no row scanning needed for pruning.
PART_GLOB_DDB = parquet_glob("partitioned").replace("\\", "/")

def q2_duckdb_part():
    return con_q.execute(f"""
        SELECT ts_date_bucketed, IncidentGrade, COUNT(*) AS alert_count
        FROM read_parquet('{PART_GLOB_DDB}', hive_partitioning=true)
        WHERE ts_date_bucketed >= '{Q2_DATES[0]}'
          AND ts_date_bucketed <= '{Q2_DATES[1]}'
        GROUP BY ts_date_bucketed, IncidentGrade
        ORDER BY ts_date_bucketed, IncidentGrade
    """).df()

benchmark_case("duckdb", "sql", "Q2_date_range", "partitioned", q2_duckdb_part)

con_q.close()
````
